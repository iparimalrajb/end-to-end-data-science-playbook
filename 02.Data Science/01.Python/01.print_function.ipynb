# Print Function

## The print() function is a built-in function in Python used to display output (data or messages) on the screen. It can take multiple arguments and can be used to format the output in various ways.

### (01) Printing a string:

```python
print("Hello World !")
```

### (02) Printing multiple values:

```python
print("Hello", "World", "," "welcome to python programming 2026")
```

### (03) Printing with a custom seperator:

```python
print("hello World", end="!!")
```

```python
print(13,"April",2026, sep="/")
```

### (04) Printing Variables:

```python
name="Raj"
age=20

print("My name is",name, "and I am", age, "years old.")
```

### (05) f-strings (formatted string ):

```python
name="Raj"
Role="Data Scientist"

print(f"My name is {name} and I am a {Role}.")
```

### (06) .format() Method:

```python
print("My name is {} and I am {} years old.".format(name,age))
```
